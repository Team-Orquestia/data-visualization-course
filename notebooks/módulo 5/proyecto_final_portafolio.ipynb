## 1. Fase 1: Generación y Preparación de Datos

Para garantizar que todos trabajemos con los mismos datos sin dependencias externas, vamos a generar un dataset sintético robusto.

```python
# EJECUTA ESTA CELDA PARA GENERAR LOS DATOS (NO MODIFICAR)
def generar_datos_ventas(n_filas=10000):
    np.random.seed(42)
    
    # Definir datos base
    fechas = pd.date_range(start='2023-01-01', end='2023-12-31', freq='h')
    categorias = ['Electrónica', 'Ropa', 'Hogar', 'Deportes']
    productos = {
        'Electrónica': ['Smartphone', 'Laptop', 'Auriculares', 'Monitor'],
        'Ropa': ['Camiseta', 'Jeans', 'Zapatillas', 'Abrigo'],
        'Hogar': ['Lámpara', 'Silla', 'Mesa', 'Decoración'],
        'Deportes': ['Pesas', 'Bicicleta', 'Balón', 'Ropa Gym']
    }
    regiones = ['Norte', 'Sur', 'Este', 'Oeste']
    tipo_cliente = ['Nuevo', 'Recurrente', 'VIP']
    
    data = []
    
    for _ in range(n_filas):
        cat = np.random.choice(categorias)
        prod = np.random.choice(productos[cat])
        fecha = np.random.choice(fechas)
        region = np.random.choice(regiones, p=[0.3, 0.2, 0.3, 0.2]) # Probabilidad sesgada
        cliente = np.random.choice(tipo_cliente, p=[0.5, 0.4, 0.1])
        
        # Lógica de precios y cantidades
        cantidad = np.random.randint(1, 10)
        precio_base = np.random.uniform(10, 500) if cat != 'Electrónica' else np.random.uniform(100, 1500)
        descuento = np.random.choice([0, 0.1, 0.2, 0.5], p=[0.7, 0.2, 0.08, 0.02])
        
        # Introducir algunos valores nulos o sucios intencionalmente
        if np.random.random() < 0.02:
            region = None
        
        data.append([fecha, cat, prod, cantidad, round(precio_base, 2), descuento, region, cliente])
        
    df = pd.DataFrame(data, columns=['Fecha', 'Categoria', 'Producto', 'Cantidad', 'Precio_Unitario', 'Descuento', 'Region', 'Tipo_Cliente'])
    
    # Ordenar por fecha
    df = df.sort_values('Fecha').reset_index(drop=True)
    return df

df_raw = generar_datos_ventas()
print("Dataset generado con éxito. Dimensión:", df_raw.shape)
df_raw.head()
```
