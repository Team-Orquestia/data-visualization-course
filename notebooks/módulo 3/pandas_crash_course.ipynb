# Crash Course de Pandas: Manipulación y Análisis de Datos

Pandas es la biblioteca fundamental para el análisis de datos en Python. Proporciona estructuras de datos flexibles y de alto rendimiento, diseñadas para trabajar con datos tabulares y etiquetados de manera fácil e intuitiva.

En este notebook cubriremos las funcionalidades esenciales:
1. Creación y carga de datos
2. Inspección básica
3. Selección y filtrado
4. Limpieza de datos (Missing values)
5. Manipulación y transformación
6. Agrupación y agregación
7. Combinación de DataFrames

```python
import pandas as pd
import numpy as np
```

## 1. Creación de Estructuras de Datos

Las dos estructuras principales en Pandas son:
*   **Series**: Un array unidimensional etiquetado (similar a una columna de Excel).
*   **DataFrame**: Una estructura tabular bidimensional (similar a una hoja de cálculo completa).

```python
# Crear un DataFrame a partir de un diccionario
data = {
    'Nombre': ['Ana', 'Juan', 'Maria', 'Pedro', 'Sofia', 'Luis'],
    'Edad': [24, 30, 28, 35, 22, 40],
    'Ciudad': ['Madrid', 'Barcelona', 'Valencia', 'Sevilla', 'Madrid', 'Barcelona'],
    'Salario': [30000, 45000, 38000, 52000, 28000, 60000],
    'Departamento': ['RRHH', 'IT', 'Ventas', 'IT', 'Marketing', 'Dirección']
}

df = pd.DataFrame(data)
df
```

## 2. Inspección de Datos

Antes de analizar, es crucial entender la estructura de nuestros datos.

```python
# Ver las primeras filas
print("--- Head ---")
print(df.head(3))

# Información general (tipos de datos, memoria, valores no nulos)
print("\n--- Info ---")
df.info()

# Estadísticas descriptivas básicas
print("\n--- Describe ---")
print(df.describe())

# Dimensiones (filas, columnas)
print(f"\nDimensiones: {df.shape}")
```

## 3. Selección y Filtrado

Pandas ofrece múltiples formas de seleccionar datos: por etiqueta (`loc`), por posición (`iloc`) o usando lógica booleana.

```python
# Selección de una columna (devuelve una Series)
print(df['Nombre'])

# Selección de múltiples columnas (devuelve un DataFrame)
subset = df[['Nombre', 'Salario']]
subset.head(2)
```

```python
# Filtrado condicional: Personas mayores de 25 años
mayores_25 = df[df['Edad'] > 25]
mayores_25
```

```python
# Filtrado con múltiples condiciones (AND '&', OR '|')
# Empleados de IT con salario mayor a 40000
filtro_it = df[(df['Departamento'] == 'IT') & (df['Salario'] > 40000)]
filtro_it
```

```python
# .loc (basado en etiquetas/nombres)
# Seleccionar filas donde la Ciudad es Madrid y solo columna Salario
print(df.loc[df['Ciudad'] == 'Madrid', 'Salario'])

# .iloc (basado en índices numéricos)
# Seleccionar filas 0 a 2 y columnas 0 a 2
print("\nSubselección por posición:")
print(df.iloc[0:3, 0:3])
```

## 4. Manipulación y Columnas Nuevas

Podemos crear nuevas variables a partir de las existentes o modificar las actuales.

```python
# Crear una nueva columna calculada
df['Salario_Mensual'] = df['Salario'] / 12

# Aplicar una función a una columna
df['Nombre_Mayus'] = df['Nombre'].apply(lambda x: x.upper())

df.head()
```

## 5. Manejo de Datos Faltantes (NaN)

Los datos del mundo real suelen estar incompletos.

```python
# Introducimos algunos valores nulos para el ejemplo
df_nan = df.copy()
df_nan.loc[0, 'Salario'] = np.nan
df_nan.loc[2, 'Ciudad'] = np.nan

print("DataFrame con nulos:")
print(df_nan)

# Detectar nulos
print("\nConteo de nulos por columna:")
print(df_nan.isna().sum())
```

```python
# Opción A: Llenar nulos (Imputación)
# Llenar salario con la media
media_salario = df_nan['Salario'].mean()
df_lleno = df_nan.fillna({'Salario': media_salario, 'Ciudad': 'Desconocido'})

# Opción B: Eliminar filas con nulos
df_limpio = df_nan.dropna()

print("DataFrame después de eliminar nulos:")
print(df_limpio)
```

## 6. Agrupación y Agregación (Groupby)

El método `groupby` es esencial para el análisis "split-apply-combine". Permite agrupar datos por categorías y calcular estadísticas.

```python
# Salario medio por Departamento
salario_por_depto = df.groupby('Departamento')['Salario'].mean().sort_values(ascending=False)
print(salario_por_depto)
```

```python
# Múltiples agregaciones
resumen = df.groupby('Ciudad').agg({
    'Salario': ['mean', 'min', 'max'],
    'Edad': 'mean',
    'Nombre': 'count' # conteo de personas
})
resumen
```

## 7. Ordenar y Valores Únicos

```python
# Ordenar por Salario descendente
df_ordenado = df.sort_values(by='Salario', ascending=False)

# Valores únicos en una columna
ciudades_unicas = df['Ciudad'].unique()

print(f"Ciudades presentes: {ciudades_unicas}")
df_ordenado.head(3)
```
